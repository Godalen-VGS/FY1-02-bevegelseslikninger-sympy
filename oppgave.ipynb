## Bevegelseslikningene

Ved konstant akselerasjon $a$, startfart $v_0$, fart $v$, tid $t$ og strekning $s$:

| Nr. | Likning | Mangler |
| --- | --- | --- |
| 1 | $v = v_0 + at$ | $s$ |
| 2 | $s = v_0 t + \frac{1}{2}at^2$ | $v$ |
| 3 | $v^2 - v_0^2 = 2as$ | $t$ |
| 4 | $s = \frac{v_0 + v}{2}\,t$ | $a$ |

Velg likningen som inneholder det du vet og det du skal finne.

## Oppskrift: fysikkoppgaver med sympy

Bruk de samme fem trinnene hver gang:

| Trinn | Hva du gjør | Sympy |
| --- | --- | --- |
| 1 | Importer bibliotekene | `import sympy as sp` og `from IPython.display import display` |
| 2 | Lag symboler for **alle** størrelsene, både de kjente og den ukjente | `v0, a, s = sp.symbols('v0 a s')` |
| 3 | Sett opp uttrykkene og likningen | `sp.Eq(venstre, høyre)` |
| 4 | Løs likningen med hensyn på den ukjente | `sp.solve(likning, s)[0]` |
| 5 | Vis formelen pent, sett inn tallene til slutt, og skriv svaret med enhet | `display(...)`, `.subs({...})` og `float(...)` |

Fordelen med å vente med tallene er at du får en **formel**. Den kan du sjekke enhetene i, bruke på nytt med andre tall, og sammenlikne med formelboka.

Gi symbolene navn som ligner det du skriver for hånd: `v0` vises som $v_0$, `v_A` som $v_A$ og `theta` som $\theta$. Se mer i **14: Sympy grunnkurs**.

## Eksempel: Bremsing

En bil kjører i 20 m/s og bremser med $a = -5$ m/s². Hvor lang er bremselengden?

Vi vet $v_0$, $v = 0$ og $a$, og skal finne $s$. Likning 3 mangler $t$, så den passer.

```python
# Trinn 1: Importere biblioteker
import sympy as sp
from IPython.display import display

# Trinn 2: Symboler for størrelsene (de kjente og den ukjente s)
v, v0, a, s, t = sp.symbols('v v0 a s t')

# Trinn 3: Sette opp likningen. Likning 3 mangler t, så den passer.
likning3 = sp.Eq(v**2 - v0**2, 2*a*s)

# Trinn 4: Løse likningen med hensyn på s
s_formel = sp.solve(likning3, s)[0]

# Trinn 5: Presentere formelen, og sette inn tallene til slutt
print("Bevegelseslikning 3:")
display(likning3)
print("Løst med hensyn på s:")
display(sp.Eq(s, s_formel))

s_svar = s_formel.subs({v: 0, v0: 20, a: -5})
print(f"Bremselengden er {float(s_svar):.3g} m")
```

```python
# Kjør denne cellen for å sjekke svarene dine. Du trenger ikke forstå koden.
def naer(a, b, rel=1e-3):
    """Sant hvis tallet a er innenfor rel (relativt) av b."""
    try:
        a, b = float(a), float(b)
    except (TypeError, ValueError):
        return False
    if b == 0:
        return abs(a) < 1e-9
    return abs(a - b) <= rel * abs(b)


def S(tekst):
    """Gjør tekst om til et sympy-uttrykk, f.eks. S("x**2 - 1")."""
    import sympy
    return sympy.sympify(tekst, locals={n: sympy.Symbol(n) for n in ("N", "O", "Q", "S")})


def lik(a, b):
    """Sant hvis a og b er matematisk like.

    Lister og mengder sammenlignes uten hensyn til rekkefølge, tupler (koordinater) i rekkefølge.
    """
    import sympy
    if a is None:
        return False
    if isinstance(b, tuple):
        if not isinstance(a, (tuple, list, sympy.Tuple)) or len(a) != len(b):
            return False
        return all(lik(x, y) for x, y in zip(a, b))
    if isinstance(b, (list, set)):
        if isinstance(a, sympy.FiniteSet):
            a = list(a)
        if not isinstance(a, (list, tuple, set)) or len(a) != len(b):
            return False
        rest = list(a)
        for y in b:
            for i, x in enumerate(rest):
                if lik(x, y):
                    del rest[i]
                    break
            else:
                return False
        return True
    if isinstance(b, dict):
        if not isinstance(a, dict) or len(a) != len(b):
            return False
        a = {str(k): v for k, v in a.items()}
        return all(str(k) in a and lik(a[str(k)], v) for k, v in b.items())
    try:
        a, b = sympy.sympify(a), sympy.sympify(b)
    except (sympy.SympifyError, TypeError):
        return False
    vanlige = lambda u: u.subs({s: sympy.Symbol(s.name) for s in u.free_symbols})
    a, b = vanlige(a), vanlige(b)
    if isinstance(a, sympy.Set) or isinstance(b, sympy.Set):
        return a == b
    if getattr(b, "is_Relational", False):
        return a == b
    try:
        return sympy.simplify(a - b) == 0
    except TypeError:
        return a == b

SJEKKER = [
    ('1a', 's_brems skal være ca. 52.1 m', 'naer(s_brems, 625/12, rel=1e-4)'),
    ('1b', 't_brems skal være ca. 4.17 s', 'naer(t_brems, 25/6, rel=1e-4)'),
    ('2a', 'h_maks skal være ca. 7.34 m', 'naer(h_maks, 144/(2*9.81), rel=1e-4)'),
    ('2b', 't_tilbake skal være ca. 2.45 s', 'naer(t_tilbake, 24/9.81, rel=1e-4)'),
    ('3', 't_uttrykk skal være de to løsningene av s = v0·t + at²/2 for t', "lik(t_uttrykk, [S('(-v0 - sqrt(v0**2 + 2*a*s))/a'), S('(-v0 + sqrt(v0**2 + 2*a*s))/a')])"),
    ('4_80', 'stopplengde(80, 1, 7) skal gi ca. 57.5 m', 'naer(stopplengde(80, 1, 7), 80/3.6 + (80/3.6)**2/14, rel=1e-4)'),
    ('4_50', 'stopplengde(50, 1.5, 8) skal gi ca. 32.9 m', 'naer(stopplengde(50, 1.5, 8), 50/3.6*1.5 + (50/3.6)**2/16, rel=1e-4)'),
    ('5', 't_innhenting skal være 20 s og s_innhenting 400 m', 'naer(t_innhenting, 20) and naer(s_innhenting, 400)'),
]


def test_alt():
    feil = 0
    for _, beskrivelse, uttrykk in SJEKKER:
        try:
            ok = eval(uttrykk)
        except Exception as e:
            ok = False
            beskrivelse += f" (krasjet: {type(e).__name__}: {e})"
        if not ok:
            print("❌", beskrivelse)
            feil += 1
    if feil == 0:
        print(f"✅ Alle testene bestått! ({len(SJEKKER)} av {len(SJEKKER)})")
    else:
        print(f"\n{len(SJEKKER) - feil} av {len(SJEKKER)} tester bestått.")


test_alt()
```
